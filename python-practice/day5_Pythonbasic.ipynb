```python
#encapsulation: keeping data and operations tht work on that data together inside a class while controlling how data is accessed

class bankAccount:
    def __init__(self,balance):
        self.balance = balance

    def deposit(self,amount):
        self.balance += amount

#python truly doesnt have private variables like java or c++ so python uses naming conventions(this is internal us hldnt directly access it)
#public the usual one     
```

```python
#protected convention: intended for internal use dont access it directly unless uk wht u are dng
class bankAccount:
    def __init__(self,balance):
        self._balance = balance

    def deposit(self,amount):
        self.balance += amount

account = bankAccount(-50000)
print(account._balance)
```

```python
#@property allows to use method like an attribute
class bankAccount:
    def __init__(self,balance):
        self._balance = balance

    @property#it tells whenuser accesses balance call this method automatically so no need to  call method 
    def balance(self):
        return self._balance

account = BankAccount(1000)

print(account.balance)
```

```python
#setter:method tht controls wht happens when u change a value
class bankAccount:
    def __init__(self, balance):
        self.balance = balance

    @property
    def balance(self):
        return self._balance

    @balance.setter
    def balance(self,value):
        if value < 0:
            print("balance cant be negative")
        else:
            self._balance = value
        
```
