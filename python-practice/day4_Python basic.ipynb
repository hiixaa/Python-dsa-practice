```python
#inheritance:way for one class to acquire attribute and methods of another class
class Animal():
    def eat(self, dish):
        self.dish = dish
class dog(Animal):
    def bark(self, types):
        self.types = types
d = dog()
d.eat("food")
d.bark("bowbow")

print(f"the dog eats {d.dish} and does {d.types}")
```

```python
#method overridiing:child class ha own verison of method even if it exists in parent class
class Animal:
    def speak(self):
        print("Animal makes a sound")


class Dog(Animal):
    def speak(self):
        print("Dog says bow bow")
        
a = Animal()
d = Dog()

a.speak()
d.speak()

```

```python
#super():used when chils class want to access smtng form parents class 
class Animal:
    def speak(self):
        print("animal makes a sound")

class dog(Animal):
    def speak(self):
        super().speak()
        print("dog barks bow bow")

a = Animal()
d = dog()

d.speak()

```

```python
#super() with init()
class Animal:
    def __init__(self, name):
        self.name = name

class Dog(Animal):
    def __init__(self,name,breed):
        self.breed = breed
        super().__init__(name)

d = Dog("bruno", "lab")

print(d.name)
print(d.breed)

#without super child overrides the parents implementation
```

```python
class Dog:
    def speak(self):
        print("Bow bow")


class Cat:
    def speak(self):
        print("Meow")


class Cow:
    def speak(self):
        print("Moo")


def make_sound(animal):#animal is empty parameter and when we say make_Sound(d) it puts d(dog object) into animal)
    animal.speak()


d = Dog()
c = Cat()
co = Cow()

make_sound(d)
make_sound(c)
make_sound(co)

#ducktyping:python cares abt wht an object can do rather than what class it belongs to

```

```python
class CreditCard:
    def pay(self,amount):
        print(f"paid${amount} using credit card")

class UPI:
    def pay(self,amount):
        print(f"paid${amount} using UPI")
class Paypal:
    def pay(self,amount):
        print(f"paid${amount} using Paypal")

def process_payment(pay_method,amount):
    pay_method.pay(amount)

c= CreditCard()
u = UPI()
p = Paypal()

process_payment(c,500)
process_payment(u,500)
process_payment(p,500)



```

```python
#abc module:abstract base class is a blueprint/rules for other classes
#It says:
#"If you inherit from me, you MUST implement certain methods."
#frm above example if every object has pay() then process_payment works for tht class  works wht if u change function name into make_payment then when u do process_oayment(cash, 500) we get error cause cash class doesnt have pay() so abc is used tht u MUST implement certain methods

from abc import ABC, abstractmethod

class Payment(ABC):
    @abstractmethod
    def pay(self, amount):
        pass

#we cant create abstract class object like p = payment()
        
class CreditCard(Payment):
    def pay(self, amount):
        print(f"Paid ${amount} using credit card")


class UPI(Payment):
    def pay(self, amount):
        print(f"Paid ${amount} using UPI")

#diff btw abstract class and inheritance the child class cant inherit ac without implementing the req method




```
