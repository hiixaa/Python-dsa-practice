```python
#init constructor:it initializes variable and gets called automatically
class Student:
    def __init__(self,name,age):
        self.name = name
        self.age = age

s1=Student("Mohitha",22)
s2 = Student("Rahul",22)

print(s1.name)
print(s1.age)

print(s2.name)
print(s2.age)
```

```python
#self means this particulare object s1
class Student:
    def __init__(self, name, age):
        self.name = name #self is s1 or s2, self.name is attribute of the obj, = name is parmeter
        self.age = age

    def introduce(self):
        print("My name is", self.name)
        print("My age is", self.age)


s1 = Student("Mohitha", 22)
s2 = Student("Rahul", 21)

s1.introduce()
s2.introduce()
```

```python
#class attributes belongs to the class itself
class Student:

    college = "ABC College" #class attribute

    def __init__(self, name, age):
        self.name = name
        self.age = age

s1 = Student("Mohitha", 22)
s2 = Student("Rahul", 21)

print(s1.college)
print(s2.college)
#both students can access college because it belogs to class not separately to eachs tudent
```

```python
#public attribute can be accessed from anywhere
class Student:
    def __init__(self, name):
        self.name = name #name is public


s1 = Student("Mohitha")

print(s1.name)
```

```python
#protected only for internal use

class Student:
    def __init__(self, name):
        self._name = name


s1 = Student("Mohitha")

print(s1._name)
```
