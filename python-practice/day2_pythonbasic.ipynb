```python
#types of arguments
#positional arg: arg tht are matched to parameters based strictly on their order
def introduce(name, age):
    print("name: ",name)
    print("age: ", age)

introduce("mohitha",22)
```

```python
#keyword arguments:arg passed by naming each parameter so it becomes order independent
def introduce(name, age):
    print("name: ",name)
    print("age: ", age)

introduce(age = 22, name = "mohitha")
#if we provide too many or too less arg then there will be a type error
```

```python
def calculate_bill(price, quantity):
    total = price*quantity
    print("total bill: ",total )

calculate_bill(100,3)
```

```python
#default arg:giving parameter a value in function definition itself
#used when its the value we need most of the time
def greet(name="Mohitha"):
    print("Hello", name)

greet()
```

```python
#default is only used when u dont provide the arg
def greet(name="Mohitha"):
    print("Hello", name)

greet("Rahul")
```

```python
#req vs defined parameters
#name-req, age-default
def introduce(name, age=22):
    print("Name:", name)
    print("Age:", age)

introduce("Mohitha")
#req parameters before default parameters else syntaxerror
```

```python
#real world example
def add_cart(product, quantity = 1):
    print(product,quantity)

add_cart("laptop")
add_cart("charger",3)
#default works with poisitonal and keyword args
```

```python
#overriding only one default
def profile(name, age = 22, city = "Vizag"):
    print(name,age,city)

profile("Mohs",city = "chennai")
```

```python
#*args:used when we donno beforehand how many values caller will provide
def add(*args):
    print(args)

add(12,15,20)
#* means collect all remaining positional args into one variable
#args collects values into tuple cause tuple is ordered and can hold multiple values and is immutable
#args can receive 0 args
```

```python
#u can have normal parameters before args
def student(name,*marks):
    print(name)
    print(marks)

    total = 0

    for mark in marks:
        total += mark

    print(total)

student("mohs",90,80,70,80)
#* in call is unpack into positional arg and in def is pack into one tuple
```

```python
#args with default arguments
def greet(message = "hello",*names):
    for name in names:
        print(message, name)

greet("Hi", "Mohitha", "Anu", "Rahul")
```

```python
#**kwargs: allows to accept any number of keyword arguments
def student_info(**kwargs):
    print(kwargs)
student_info(name="Mohitha", age=22, city="Vizag")
#all these args are collected into a dictionary
```

```python
#looping through kwargs 
def student_info(**kwargs):

    for key, value in kwargs.items():
        print(key, "=", value)

student_info(
    name="mohtiha",
    age = 22,
    city = "vizag"
)
#btw nrml arg dict and kwargs the diff is in kwargs dict is created for u
#order of parameters def function(normal, *args, **kwargs):
```
