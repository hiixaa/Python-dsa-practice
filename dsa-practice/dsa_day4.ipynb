```python
#recursion basics
#recursion is when a function calls itself to solve a smaller version of same problem
def count(n):
    print(n)
    count(n - 1)#here it calls count(2)

count(3)
#prblm is it doesnt know when to stop so base case comes in

#base case:conditions which tells recursive funtion when to stop calling itself

def count(n):
    if n == 0:#checks if we reached stopping point #base case
        return#exits from function immediately

    print(n)
    count(n-1)#recursive case

count(3)

#HOW DOES PYTHON REMemBERS FUNCTION CALLS
# every time recursive function called itself python keep track of current function call
# this is managed using call stack
```

```python
#FACTORIAL USING RECURSION
#factorial using looops
n = 5
fact =1

for i in range(1, n+1):
    fact = fact*i
print(fact)

#RECURSIVE WAY
# n! = n * (n-1)! ----> recursive case
# 0!=1 or n==0 -----> base case
def factorial(n):
    if n<0:
        return"factorial is not defined for negative numbers"
    if n==0:
        return 1

    return n*factorial(n-1)

print(factorial(5))
```

```python
#fibonacci sequence
# fib(n) = fib(n-1)+fib(n-2) ----->using recursion

def fib(n):
    if n ==0:#base case
        return 0

    if n== 1:#base case
        return 1

    return fib(n-1)+fib(n-2) #recursive case

print(fib(5))
```

```python
#sum of digits using recursion
# - n % 10 → extracts the last digit.
# - n // 10 → removes the last digit.
# n == 0 return 0 ----> base case
# sum(1234) = 4+sum(123) ---> recursive case

def sum_digits(n):
    n=abs(n)#supports negative numbers
    if n==0:
        return 0

    return (n%10) + sum_digits(n//10)

print(sum_digits(1234))
```
