```python
#DUPLICATE DETECTION IN PYTHON (HASHING):THIS IS BASICALLY WITHOUT USING SETS OR DICTIOARY
#hashing: 
# suppose u want to check a number exists or not u can check if num == 45 found,it can check all numbers 
# now if u want to check in millions of numbers its not efficient

# hashing is a technique tht uses hash function to convert value into hash value which helsp ds to locate and organize effiviently

#DUPLICATE DETECTION IN PYTHON
#checking if its duplicate
def contains_duplicate(numbers):
    seen = set()

    for num in numbers:
        if num in seen:
            return True

        seen.add(num)

    return False
print(contains_duplicate([2,3,3,3,4,5,6,7,6,7,6,7]))

#find all duplicate elements

def find_duplicates(numbers):
    seen = set()
    duplicates = set()

    for num in numbers:
        if num in seen:
            duplicates.add(num)
        else:
            seen.add(num)

    return list(duplicates)

print(find_duplicates([2,3,3,3,4,5,6,7,6,7,6,7]))

#count the occurences of each element
def freq_count(numbers):
    frequency = {}

    for num in numbers:
        frequency[num] = frequency.get(num,0) +1
    return frequency

print(freq_count([1,2,3,2,1,2]))

#find first repeated element

def first_repeated(numbers):
    seen = set()

    for num in numbers:
        if num in seen:
            return num

        seen.add(num)
    return -1

print(first_repeated([5, 3, 4, 3, 5, 6]))
```

```python
#2sum brute force o(n2)
def two_sum(numbers, target):
    for i in range(len(numbers)):
        for j in range(i + 1, len(numbers)):
            if numbers[i] + numbers[j] == target:
                return [i, j]

    return []

#2sum hashing
def t_sum(numbers,target):
    seen = {}

    for i, num in enumerate(numbers):
        complement = target - num

        if complement in seen:
            return[seen[complement],i]

        seen[num]=i

    return []

print(t_Sum([2,7,8,6], 9))
```

```python
#grouping patterns using dictionaries
fruits = ["apple", "banana", "apricot", "blueberry", "avocado"]

groups = {}

for fruit in fruits:
    key = fruit[0]

    if key not in groups:
        groups[key] = []

    groups[key].append(fruit)

print(groups)

#set default method

fruits = ["apple", "banana", "apricot", "blueberry", "avocado"]

groups = {}

for fruit in fruits:
    key = fruit[0]
    groups.setdefault(key, []).append(fruit)

print(groups)
```
