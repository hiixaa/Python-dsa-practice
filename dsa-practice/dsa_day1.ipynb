```python
#2 pointer technique used to solve array and string problems
#pointer:variable tht stores index or position in an string

#instead of using one index to process elements we use 2 to compare elements

def reverse_array(numbers):
    left = 0
    right = len(numbers) - 1
    #reverse and array
    while left < right:
        #so swap elements
        numbers[left], numbers[right] = numbers[right], numbers[left]
        
        #move pointers
        left +=1
        right -= 1

    return numbers

print(reverse_array([10,20,30,40,50]))
#o(n) and auxilary space complaxity :o(1)
#nrml for looping o(n2),because of shifting elmenets evrytime
numbers = [10, 20, 30, 40, 50]
reversed_numbers = []

for num in numbers:
    reversed_numbers.insert(0, num)
```

```python
#palindrome
def is_palindrome(word):
    left = 0
    right = len(word) - 1

    while left  <  right:
        if word[left] != word[right]:
            return False

        left += 1
        right -= 1

    return True

print(is_palindrome("madam"))

#o(n) for 6 - 3 for n n/2 so n
#auxiliary space o(1) 
#we used fixed number of extra variables irrespective of no of elements
```

```python
#binary search faster on large collections but list must be sorted

def binary_Search(numbers, target):
    left = 0
    right = len(numbers)-1

    while left <= right:
        mid = (left + right) //2

        if numbers[mid] == target:
            return mid
        elif numbers[mid] < target: #discarding the remaining part including mid and moving the left pointer to mid
            left = mid + 1
        else:
            right = mid -1

    return -1 #if target exists it returns index or it returns -1

numbers = [10,20,30,40,50,60,70]

print(binary_Search(numbers, 60))
#O(LOGN)
```

```python
#SORTING ALGOS
#BUBBLE SORT : compares 2 adjacent elements and swaps them if they are in wrong order
#after one full iteration(pass) on all elements then do again
numbers = [5,2,8,1,3]

n = len(numbers)


for i in range (n):#how many passes we perform
    for j in range(n - 1 - i):#(compares neightbouring elements)(n-1-i reduces comparisions because each pass places one more largest element at the end)
        if numbers[j] > numbers[j + 1]:
            numbers[j], numbers[j+1] = (numbers[j + 1], numbers[j])
print(numbers)

#worst case: o(n2) since algo makes n in first pass, n-1...total grows propotionally to n2
#space: o(1)
```

```python
#SELECTION SORT repeatedly finds smallest element in unsorted part and places it at beginning of tht part)
#first strt from start and compare first number with all if it is bigger than other number go to the other number and when u find least replace its position to one
numbers = [5,2,8,1,3]

n= len(numbers)

for i in range(n -1):
    min_index = i#initially assume min index is first one

    for j in range(i + 1, n):
        if numbers[j] < numbers[min_index]:
            min_index = j

    numbers[i], numbers[min_index] = (numbers[min_index], numbers[i])
print (numbers)
#space:o(1)
#time:o(n2)
```

```python
#INSERTION SORT BUILDS SORTED SECTION ONE AT A TIME BY INSERTING EACH NEW ELEMENT INTO ITS CORRECT POSITION
#like after one number key is sorted it goes to another
numbers = [5,2,4,1,3]

for i in range(1, len(numbers)):#i is index of element we want to insert into sorted section
    key = numbers[i]#element to insert
    j = i -1#index of prev element

    #shift larger elements one position to right
    while j >= 0 and numbers[j] > key:
        numbers[j+1] = numbers[j]#shift larger elements to right
        j -= 1
#key stores element we want to insert
#j starts at index to its left
    #insert key
    numbers[j+1] = key

print(numbers)
#o(n2),
#space:o(1)
```
